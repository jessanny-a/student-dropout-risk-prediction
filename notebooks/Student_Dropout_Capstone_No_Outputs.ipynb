# Predicting Student Dropout Risk Using Explainable and Fair Machine Learning: An Early-Warning System for Proactive Academic Intervention

## Step 1: Problem Understanding and Framing

### Business Problem

Student dropout creates significant consequences for both students and educational institutions. Students who leave before completing their studies may lose time, tuition investment, and future career opportunities. At the same time, institutions may experience lower completion rates, lost tuition revenue, inefficient use of academic resources, and reputational impact.

Academic support resources such as advisers, tutoring programs, financial assistance, and counseling are also limited. Institutions therefore need a systematic way to identify students who may require additional support before dropout occurs.

This project develops a machine-learning early-warning system that estimates each student's risk of dropping out. The model is intended to help academic advisers **prioritize proactive interventions**, not to make automatic decisions about students.

**Business problem: Can student information available early in the academic journey be used to identify students at elevated risk of dropping out accurately, fairly, and early enough for the institution to intervene?**

### Data Science Problem

In this project, a **supervised binary classification** data science task will be used.

A classification task is appropriate because the business outcome is categorical: whether a student ultimately **drops out** or **graduates**. However, the goal is not simply to assign a class. The model should produce a **dropout-risk probability**.

This probability lets the institution rank students by risk and decide who should receive additional review or intervention based on available resources.

If academic advisers have the capacity to proactively contact 15-20% of students, the institution could prioritize those with the highest predicted risk.

Records whose final status remains “Enrolled” will be excluded from the primary binary experiment to avoid ambiguous labels and may be used in a secondary multiclass analysis.

The model will estimate:

$$ P(\text{Dropout} \mid X) $$

where \(X\) represents information available about a student before the final academic outcome is known.

The primary target will be:

1 = Dropout\
0 = Graduate


### Early Warning Design

In this project, we will compare two prediction models.

#### Model A: Enrollment-Time Risk

This model uses only information that is available when or near the time a student enters the institution.

Examples include:
* Demographic characteristics
* Prior qualifications
* Admission information
* Scholarship indicators
* Socioeconomic variables
* Course / Program Information

This answers the business question: **How much dropout risk can we identify before the student begins accumulating university academic results?**

#### Model B: End of first Semester Risk

This model adds first semester academic information.

This answers the business question: **How much predictive improvement is gained by waiting until first semester performance becomes available?**

Comparing the two models, gives us a meaningful business tradeoff: **Earlier prediction versus more accurate prediction.**

Second-semester variables will be excluded from the operational models to preserve early-warning usefulness and reduce temporal leakage.

### Stakeholders

The primary stakeholders are:
* **Students**: may receive earlier academic, financial, or personal support.
* **Academic advisers and student-success teams**: use the model as decision support for prioritizing outreach
* **Faculty and program administrators**: gain insight into patterns associated with student attrition
* **Institutional leadership**: evaluates potential improvement in retention, completion, and resource allocation.
* **Data/IT teams**: responsible for maintaining the model, data pipeline, monitoring, access control, and governance.

**Note:** The model does not independently determine sanctions, admission status, scholarships, or student eligibility. Its role is risk identification and prioritization for human review.

### Success Metrics

#### Primary Metric

The primary metric we are going to use is **Recall for the Dropout Class**.

Recall = TP / (TP + FN)

* **TP (True Positives)**: The number of actual positive instances that the model correctly predicted as positive.
* **FN (False Negatives)**: The number of actual positive instances that the model mistakenly predicted as negative.
* **TP + FN**: This sum represents the total number of actual positive instances present in your data, regardless of what the model predicted.

This measures: Of all students who actually drop out, what percentage did the model identify as being at risk?

This is important because a false negative represents an at-risk student whom the system fails to flag for potential support.

#### Secondary Metric

We will also report:
* **ROC-AUC**: Overall ability to distinguish dropout from graduate students across thresholds
* **Precision**: of students flagged as high-risk, how many actually belong to the dropout group
* **F1-score**: Balance between precision and recall.
* **PR-AUC**: Useful if the target classes are meaningfully imbalanced.
* **Balanced Accuracy**: Reduces the influence of unequal class sizes.
* **Confusion Matrix**: Communicates false positives and false negatives directly.

### Business Key Performance Indicators

1. **At-Risk Student Capture Rate** - Percentage of eventual dropouts contained within the group selected for intervention
2. **Precision of Intervention** - Percentage of students for intervention who are actually high-risk cases
3. **Intervention Workload** - Number or percentage of students requiring adviser review under a chosen risk threshold.
4. **Potential Retention Improvement** - This identifies the students who may be candidates for support. Note that this does not establish that an intervention will prevent dropout.

   Potential Retained Students = Flagged True Dropouts × Assumed Intervention Effectiveness

5. **Potential Financial Value**

   Potential Financial Value= Additional Retained Students × Estimated Tuition/Student − Intervention Cost

   Because the dataset does not provide the institution's tuition or intervention costs, these are just scenario assumptions rather than observed facts.

### Responsible-AI Objective:

Evaluate model explanations and error rates across relevant sensitive groups, identify possible disparities, and propose feasible mitigation strategies while maintaining human oversight.

### Data Quality Assessment

Independent validation of the supplied CSV identified no missing values and no exact duplicate observations. All predictor variables are physically represented numerically; however, several variables—including Course, Application mode, Nationality and parental qualifications and occupations—are coded categorical variables and will therefore not be treated as continuous measurements. A minor formatting issue was identified in the Daytime/evening attendance column name, which contains trailing whitespace and will be corrected during preprocessing. Preliminary IQR-based screening identified statistical extremes in variables including age, academic grades and curricular-unit counts. These observations will not be removed automatically, since many appear to represent plausible student characteristics rather than data errors. The project will validate outliers using domain plausibility before deciding on any treatment.

### Important Data Understanding Objectives

1. Integer does not necessarily mean numeric. Course, Nationality, Application mode, parent occupations, and qualifications are stored as numbers but are categories.
2. The source contains no reported missing values, but some category mappings contain concepts such as "Unknown" or "Other". Those are valid encoded categories and should not automatically be converted to NaN.
3. The dataset has target imbalance. 'Graduate' is the largest class and 'Enrolled' is the smallest. This reinforces our decision not to judge model quality using accuracy alone.
4. Academic variables are strongly time-dependent. Using Semester 2 variables in a model advertised as an early Semester 1 intervention system would create an operational mismatch even if it increases model accuracy.
5. Sensitive attributes exist. Gender, age, nationality/international status, and socioeconomic-related attributes will require careful treatment in the analysis.

### Data Limitations

The data came from a particular higher-education setting in Portugal, so results may not generalize directly to universities in the Philippines or other educational systems.

The target describes observed outcomes, but the dataset does not establish why individual students dropped out. Prediction should therefore not be interpreted as casual inference.

Some variables - such as tuition status, debtor status and parental characteristics - may act as socioeconomic proxies. Predictive usefulness does not automatically justify their unrestrictred operational use.

Lastly, the source dataset has already undergone preprocessing which is convenient. THerefore, this capstone will clearly distinguish source preprocessing versus my own reproducible validation and transformation.

### Clean Column Names

```python
# ============================================================
# CLEAN COLUMN NAMES
# ============================================================

def clean_column_name(column):
    """
    Convert column names to lowercase snake_case.
    Removes leading/trailing whitespace and special characters.
    """
    
    column = column.strip()
    column = column.lower()
    
    column = re.sub(r"[()]", "", column)
    column = re.sub(r"['’]", "", column)
    column = re.sub(r"[/\-\s]+", "_", column)
    column = re.sub(r"[^a-z0-9_]", "", column)
    column = re.sub(r"_+", "_", column)
    
    return column.strip("_")


df = df_raw.copy()

df.columns = [
    clean_column_name(col)
    for col in df.columns
]

# Fix original UCI spelling
df = df.rename(columns={
    "nacionality": "nationality"
})

print(df.columns.tolist())
```

### Outlier Analysis

#### IQR outlier function

```python
# ============================================================
# OUTLIER ASSESSMENT
# ============================================================

def iqr_outlier_summary(data, columns):
    
    results = []
    
    for col in columns:
        
        q1 = data[col].quantile(0.25)
        q3 = data[col].quantile(0.75)
        
        iqr = q3 - q1
        
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        
        mask = (
            (data[col] < lower)
            | (data[col] > upper)
        )
        
        results.append({
            "feature": col,
            "q1": q1,
            "q3": q3,
            "iqr": iqr,
            "lower_fence": lower,
            "upper_fence": upper,
            "outlier_count": mask.sum(),
            "outlier_percent": mask.mean() * 100
        })
    
    return pd.DataFrame(results)
```

```python
outlier_summary["treatment"] = "Retain — valid/plausible observation"

outlier_summary.loc[
    outlier_summary["iqr"] == 0,
    "treatment"
] = "Retain — zero-inflated; IQR rule not appropriate"

outlier_summary.sort_values(
    "outlier_percent",
    ascending=False
)
```

```python
outlier_summary.to_csv(
    REPORTS / "outlier_summary.csv",
    index=False
)
```

Outlier Treatment. Potential outliers were identified using the 1.5×IQR rule. Several variables produced relatively high outlier counts; however, further inspection indicated that many represented plausible observations rather than erroneous data. For example, students above age 34 were statistically flagged despite representing legitimate mature learners, while zero-inflated curricular-unit variables produced an IQR of zero, causing any positive observation to be classified as an outlier. Similarly, low or zero semester grades may represent genuine academic non-performance and are directly relevant to the prediction task. Consequently, no observations were removed solely on the basis of the IQR rule. Valid extreme values were retained to preserve information and avoid introducing selection bias.

### Feature Engineering

#### Safe Division Function

```python
# ============================================================
# FEATURE ENGINEERING
# ============================================================

def safe_divide(numerator, denominator):
    
    return np.where(
        denominator > 0,
        numerator / denominator,
        0
    )
```

#### Domain-derived features

```python
df_fe = df_binary.copy()


# ------------------------------------------------------------
# ENROLLMENT-TIME FEATURES
# ------------------------------------------------------------

# Difference between admission performance and previous
# qualification performance
df_fe["academic_preparedness_gap"] = (
    df_fe["admission_grade"]
    - df_fe["previous_qualification_grade"]
)


# Average measure of prior academic preparation
df_fe["academic_preparedness_mean"] = (
    df_fe[
        [
            "admission_grade",
            "previous_qualification_grade"
        ]
    ]
    .mean(axis=1)
)


# Simple financial-risk indicator
# 0 = neither risk condition
# 1 = one risk condition
# 2 = debtor AND tuition not current

df_fe["financial_risk_count"] = (
    df_fe["debtor"]
    + (1 - df_fe["tuition_fees_up_to_date"])
)


# ------------------------------------------------------------
# FIRST-SEMESTER FEATURES
# ------------------------------------------------------------

df_fe["sem1_completion_rate"] = safe_divide(
    df_fe["curricular_units_1st_sem_approved"],
    df_fe["curricular_units_1st_sem_enrolled"]
)


df_fe["sem1_evaluation_success_rate"] = safe_divide(
    df_fe["curricular_units_1st_sem_approved"],
    df_fe["curricular_units_1st_sem_evaluations"]
)


df_fe["sem1_without_evaluation_rate"] = safe_divide(
    df_fe[
        "curricular_units_1st_sem_without_evaluations"
    ],
    df_fe["curricular_units_1st_sem_enrolled"]
)


df_fe["sem1_credit_rate"] = safe_divide(
    df_fe["curricular_units_1st_sem_credited"],
    df_fe["curricular_units_1st_sem_enrolled"]
)


print("Feature-engineered dataset shape:")
print(df_fe.shape)
```

#### Preprocessing Function

```python
# ============================================================
# PREPROCESSING PIPELINE
# ============================================================

def create_preprocessor(
    categorical_columns,
    numeric_columns
):
    
    categorical_pipeline = Pipeline(
        steps=[
            (
                "imputer",
                SimpleImputer(
                    strategy="most_frequent"
                )
            ),
            (
                "onehot",
                OneHotEncoder(
                    handle_unknown="ignore"
                )
            )
        ]
    )
    
    
    numeric_pipeline = Pipeline(
        steps=[
            (
                "imputer",
                SimpleImputer(
                    strategy="median"
                )
            ),
            (
                "scaler",
                StandardScaler()
            )
        ]
    )
    
    
    preprocessor = ColumnTransformer(
        transformers=[
            (
                "cat",
                categorical_pipeline,
                categorical_columns
            ),
            (
                "num",
                numeric_pipeline,
                numeric_columns
            )
        ]
    )
    
    return preprocessor
```

Principal Component Analysis (PCA) was applied to the standardized numerical predictors in the end-of-first-semester feature set to evaluate whether the feature space could be reduced while retaining most of its information. PCA was fitted exclusively on the training data to avoid information leakage. The analysis included 28 numerical and binary-derived predictors.

The first principal component explained approximately 17.83% of total variance, while the first two components together explained 30.44%. Variance was distributed across multiple dimensions rather than being dominated by a small number of components. Twelve principal components were required to retain approximately 82.33% of the variance, 15 components retained 90.35%, and 18 components retained 96.20%.

Therefore, a 90%-variance PCA representation could reduce the numerical feature space from 28 variables to 15 principal components, corresponding to approximately a 46% reduction in dimensionality. However, this reduction comes at the cost of interpretability because each principal component represents a weighted combination of multiple original variables.

The final principal components contributed little or no additional variance, indicating redundancy among some of the original and engineered features. This was expected because several engineered variables were mathematically derived from existing predictors, such as academic-preparedness measures and the financial-risk indicator.

PCA was therefore retained as a dimensionality-reduction experiment and will be evaluated against the original interpretable feature set during model development. PCA will not automatically be adopted for the final model unless it provides a meaningful performance or computational advantage that justifies the corresponding loss of interpretability.

#### Pipeline Builder

```python
# ============================================================
# PIPELINE BUILDER
# ============================================================

def build_model_pipeline(
    model,
    categorical_columns,
    numeric_columns
):
    
    preprocessor = create_preprocessor(
        categorical_columns,
        numeric_columns
    )
    
    
    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", clone(model))
        ]
    )
    
    return pipeline
```

#### Cross-valdiation function

```python
# ============================================================
# CROSS-VALIDATION FUNCTION
# ============================================================

def evaluate_models_cv(
    X,
    y,
    models,
    categorical_columns,
    numeric_columns,
    horizon_name
):
    
    results = []
    
    
    for model_name, model in models.items():
        
        print(
            f"Evaluating {horizon_name} — {model_name}..."
        )
        
        
        pipeline = build_model_pipeline(
            model=model,
            categorical_columns=categorical_columns,
            numeric_columns=numeric_columns
        )
        
        
        start_time = time.time()
        
        
        cv_results = cross_validate(
            estimator=pipeline,
            X=X,
            y=y,
            cv=cv_strategy,
            scoring=scoring,
            return_train_score=False,
            n_jobs=-1
        )
        
        
        elapsed = time.time() - start_time
        
        
        row = {
            "horizon": horizon_name,
            "model": model_name,
            "accuracy_mean":
                cv_results["test_accuracy"].mean(),
            "accuracy_std":
                cv_results["test_accuracy"].std(),
            
            "balanced_accuracy_mean":
                cv_results[
                    "test_balanced_accuracy"
                ].mean(),
            
            "precision_mean":
                cv_results["test_precision"].mean(),
            
            "recall_mean":
                cv_results["test_recall"].mean(),
            
            "f1_mean":
                cv_results["test_f1"].mean(),
            
            "roc_auc_mean":
                cv_results["test_roc_auc"].mean(),
            
            "roc_auc_std":
                cv_results["test_roc_auc"].std(),
            
            "pr_auc_mean":
                cv_results["test_pr_auc"].mean(),
            
            "fit_time_seconds":
                cv_results["fit_time"].mean(),
            
            "total_elapsed_seconds":
                elapsed
        }
        
        
        results.append(row)
    
    
    return pd.DataFrame(results)
```

### Threshold strategy

### Probability-Threshold Optimization Strategy

Classification algorithms typically convert predicted probabilities into
class labels using a default threshold of 0.50. However, the optimal threshold
depends on the business objective.

For this project, missing a student who eventually drops out (false negative)
may be more costly than flagging a student who ultimately graduates
(false positive). Therefore, threshold optimization focuses particularly on
dropout recall while also monitoring precision, F1-score, balanced accuracy,
and the proportion of students requiring intervention.

Threshold selection is performed exclusively using out-of-fold predictions
from the training dataset. The final holdout test set remains untouched.

### Threshold Evaluation

### Create Threshold Metric Function

To calculate operational metrics at any threshold

```python
# ============================================================
# THRESHOLD EVALUATION FUNCTION
# ============================================================

def evaluate_threshold(
    y_true,
    probabilities,
    threshold
):

    predictions = (
        probabilities >= threshold
    ).astype(int)


    tn, fp, fn, tp = confusion_matrix(
        y_true,
        predictions
    ).ravel()


    precision = precision_score(
        y_true,
        predictions,
        zero_division=0
    )


    recall = recall_score(
        y_true,
        predictions,
        zero_division=0
    )


    f1 = f1_score(
        y_true,
        predictions,
        zero_division=0
    )


    f2 = fbeta_score(
        y_true,
        predictions,
        beta=2,
        zero_division=0
    )


    balanced_accuracy = (
        balanced_accuracy_score(
            y_true,
            predictions
        )
    )


    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else np.nan
    )


    false_positive_rate = (
        fp / (fp + tn)
        if (fp + tn) > 0
        else np.nan
    )


    alert_rate = predictions.mean()


    return {

        "threshold": threshold,

        "precision": precision,

        "recall": recall,

        "f1": f1,

        "f2": f2,

        "balanced_accuracy":
            balanced_accuracy,

        "specificity":
            specificity,

        "false_positive_rate":
            false_positive_rate,

        "alert_rate":
            alert_rate,

        "true_negatives":
            tn,

        "false_positives":
            fp,

        "false_negatives":
            fn,

        "true_positives":
            tp
    }
```

### Why F2?

F1 weights precision and recall equally.

F2 gives greater weight to recall, which fits our business objective of minimizing missed at-risk students.

### Create Threshold Sweep Function

We'll test thresholds from 0.05 to 0.95.

```python
# ============================================================
# THRESHOLD SWEEP FUNCTION
# ============================================================

def threshold_sweep(
    y_true,
    probabilities
):

    thresholds = np.arange(
        0.05,
        0.951,
        0.01
    )


    rows = []


    for threshold in thresholds:

        rows.append(
            evaluate_threshold(
                y_true=y_true,
                probabilities=probabilities,
                threshold=threshold
            )
        )


    return pd.DataFrame(rows)
```

For example, a threshold of 0.05 produced recall of approximately 98.4%, but precision was only 44.8% and the false-positive rate reached approximately 78.1%. As the threshold increased, precision and specificity improved while recall declined gradually. This demonstrates why the default 0.50 cutoff should not automatically be assumed to be optimal and why threshold selection must balance student detection with intervention capacity.

### Identify Candidate Threshold

### Function to identify threshold candidates

```python
# ============================================================
# THRESHOLD CANDIDATE SELECTION
# ============================================================

def identify_threshold_candidates(
    threshold_df,
    model_name,
    target_recall=0.90
):

    candidates = []


    # ----------------------------------------
    # Default 0.50 threshold
    # ----------------------------------------

    default_row = (
        threshold_df.iloc[
            (
                threshold_df["threshold"]
                - 0.50
            ).abs().argsort()[:1]
        ]
        .iloc[0]
    )


    candidates.append({
        "model": model_name,
        "strategy": "Default 0.50",
        **default_row.to_dict()
    })


    # ----------------------------------------
    # Maximum F1
    # ----------------------------------------

    best_f1 = threshold_df.loc[
        threshold_df["f1"].idxmax()
    ]


    candidates.append({
        "model": model_name,
        "strategy": "Max F1",
        **best_f1.to_dict()
    })


    # ----------------------------------------
    # Maximum F2
    # ----------------------------------------

    best_f2 = threshold_df.loc[
        threshold_df["f2"].idxmax()
    ]


    candidates.append({
        "model": model_name,
        "strategy": "Max F2",
        **best_f2.to_dict()
    })


    # ----------------------------------------
    # Maximum balanced accuracy
    # ----------------------------------------

    best_balanced = threshold_df.loc[
        threshold_df[
            "balanced_accuracy"
        ].idxmax()
    ]


    candidates.append({
        "model": model_name,
        "strategy":
            "Max Balanced Accuracy",
        **best_balanced.to_dict()
    })


    # ----------------------------------------
    # Highest precision while recall >= target
    # ----------------------------------------

    target_rows = threshold_df[
        threshold_df["recall"]
        >= target_recall
    ]


    if len(target_rows) > 0:

        target_row = target_rows.loc[
            target_rows[
                "precision"
            ].idxmax()
        ]


        candidates.append({
            "model": model_name,
            "strategy":
                f"Recall >= {target_recall:.0%}",
            **target_row.to_dict()
        })


    return pd.DataFrame(candidates)
```

#### Holdout evaluation function

```python
# ============================================================
# FINAL HOLDOUT EVALUATION FUNCTION
# ============================================================

def evaluate_holdout(
    y_true,
    probabilities,
    threshold,
    model_name,
    policy
):

    predictions = (
        probabilities >= threshold
    ).astype(int)


    tn, fp, fn, tp = confusion_matrix(
        y_true,
        predictions,
        labels=[0, 1]
    ).ravel()


    accuracy = accuracy_score(
        y_true,
        predictions
    )


    balanced_accuracy = balanced_accuracy_score(
        y_true,
        predictions
    )


    precision = precision_score(
        y_true,
        predictions,
        zero_division=0
    )


    recall = recall_score(
        y_true,
        predictions,
        zero_division=0
    )


    f1 = f1_score(
        y_true,
        predictions,
        zero_division=0
    )


    f2 = fbeta_score(
        y_true,
        predictions,
        beta=2,
        zero_division=0
    )


    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else np.nan
    )


    roc_auc = roc_auc_score(
        y_true,
        probabilities
    )


    pr_auc = average_precision_score(
        y_true,
        probabilities
    )


    alert_rate = predictions.mean()


    return {

        "model": model_name,

        "policy": policy,

        "threshold": threshold,

        "accuracy": accuracy,

        "balanced_accuracy":
            balanced_accuracy,

        "precision": precision,

        "recall": recall,

        "f1": f1,

        "f2": f2,

        "specificity": specificity,

        "roc_auc": roc_auc,

        "pr_auc": pr_auc,

        "alert_rate": alert_rate,

        "true_negatives": tn,

        "false_positives": fp,

        "false_negatives": fn,

        "true_positives": tp
    }
```

#### SHAP Reconstruction QA

Verify that SHAP reconstructs the model probability

For XGBoost binary classification:

$$ P(\text{Dropout}) = \sigma( \text{base value} + \sum SHAP_i ) $$

```python
# ============================================================
# SHAP RECONSTRUCTION QA
# ============================================================

reconstructed_margin = (
    shap_base_values
    +
    shap_values.sum(axis=1)
)


reconstructed_probability = expit(
    reconstructed_margin
)


max_probability_difference = np.max(
    np.abs(
        reconstructed_probability
        -
        xgb_test_proba
    )
)


print(
    "Maximum SHAP probability reconstruction difference:"
)

print(
    max_probability_difference
)


assert max_probability_difference < 1e-4


print(
    "\nSHAP probability reconstruction passed."
)
```

#### Map Encoded Features back to Raw Variables

```python
# ============================================================
# MAP TRANSFORMED FEATURES BACK TO RAW VARIABLES
# ============================================================

raw_feature_names = list(
    X_B_train.columns
)


# Sort longest first to avoid partial-name collisions
raw_feature_names_sorted = sorted(
    raw_feature_names,
    key=len,
    reverse=True
)


def map_to_raw_feature(
    transformed_name
):

    # Remove ColumnTransformer prefix such as:
    # num__age_at_enrollment
    # cat__course_9500

    if "__" in transformed_name:

        clean_name = (
            transformed_name
            .split("__", 1)[1]
        )

    else:

        clean_name = transformed_name


    # Exact match first
    if clean_name in raw_feature_names:

        return clean_name


    # Match one-hot encoded categorical columns
    for raw_feature in raw_feature_names_sorted:

        prefix = (
            raw_feature
            + "_"
        )

        if clean_name.startswith(prefix):

            return raw_feature


    # Fallback
    return clean_name
```

#### Local explanation helper

```python
# ============================================================
# LOCAL SHAP EXPLANATION FUNCTION
# ============================================================

def explain_holdout_record(
    row_position,
    top_n=12
):

    record_shap = shap_values[
        row_position
    ]


    contribution_table = pd.DataFrame({

        "transformed_feature":
            transformed_feature_names,

        "shap_contribution":
            record_shap
    })


    contribution_table[
        "raw_feature"
    ] = (
        contribution_table[
            "transformed_feature"
        ]
        .apply(
            map_to_raw_feature
        )
    )


    # Aggregate one-hot variables back to raw feature
    local_raw = (
        contribution_table
        .groupby(
            "raw_feature",
            as_index=False
        )
        ["shap_contribution"]
        .sum()
    )


    local_raw[
        "absolute_contribution"
    ] = (
        local_raw[
            "shap_contribution"
        ]
        .abs()
    )


    local_raw[
        "direction"
    ] = np.where(

        local_raw[
            "shap_contribution"
        ] > 0,

        "Increases predicted dropout risk",

        "Decreases predicted dropout risk"
    )


    # Add raw input values where available
    original_row = (
        X_B_test
        .iloc[row_position]
    )


    local_raw[
        "raw_value"
    ] = (
        local_raw[
            "raw_feature"
        ]
        .map(
            original_row.to_dict()
        )
    )


    local_raw = (
        local_raw
        .sort_values(
            "absolute_contribution",
            ascending=False
        )
        .head(top_n)
        .reset_index(drop=True)
    )


    return local_raw
```

## Step 5B: Fairness & Bias Auditing

### Fairness and Bias Audit

The final XGBoost model is evaluated for subgroup performance using the frozen
0.48 decision threshold. No model parameters or thresholds are changed based
on the fairness results.

The audit examines:

- subgroup sample size and dropout prevalence,
- intervention / alert rate,
- precision,
- recall (true-positive rate),
- false-negative rate,
- specificity,
- false-positive rate,
- balanced accuracy,
- ROC-AUC and PR-AUC where estimable, and
- predicted-risk calibration by subgroup.

Particular attention is given to false-negative disparities because students
who are not identified as at risk may not receive proactive intervention.

Group differences are treated as diagnostic evidence rather than automatic
proof of unfairness. Differences may reflect sample size, underlying outcome
prevalence, measurement quality, historical inequities, or model behavior and
therefore require contextual interpretation.

### Subgroup Fairness Metric Function

```python
# ============================================================
# SUBGROUP FAIRNESS METRIC FUNCTION
# ============================================================

def safe_divide(
    numerator,
    denominator
):

    if denominator == 0:
        return np.nan

    return (
        numerator
        /
        denominator
    )


def subgroup_metrics(
    data,
    group_column
):

    results = []


    for group_value, group_data in data.groupby(
        group_column,
        observed=False,
        dropna=False
    ):

        y_true = (
            group_data[
                "actual_dropout"
            ]
            .astype(int)
            .to_numpy()
        )


        y_pred = (
            group_data[
                "predicted_dropout"
            ]
            .astype(int)
            .to_numpy()
        )


        y_prob = (
            group_data[
                "dropout_probability"
            ]
            .astype(float)
            .to_numpy()
        )


        tn, fp, fn, tp = (
            confusion_matrix(
                y_true,
                y_pred,
                labels=[0, 1]
            )
            .ravel()
        )


        n = len(group_data)

        actual_dropouts = (
            tp + fn
        )

        actual_graduates = (
            tn + fp
        )


        dropout_prevalence = (
            y_true.mean()
            if n > 0
            else np.nan
        )


        alert_rate = (
            y_pred.mean()
            if n > 0
            else np.nan
        )


        precision = safe_divide(
            tp,
            tp + fp
        )


        recall = safe_divide(
            tp,
            tp + fn
        )


        false_negative_rate = (
            safe_divide(
                fn,
                fn + tp
            )
        )


        specificity = safe_divide(
            tn,
            tn + fp
        )


        false_positive_rate = (
            safe_divide(
                fp,
                fp + tn
            )
        )


        accuracy = (
            accuracy_score(
                y_true,
                y_pred
            )
            if n > 0
            else np.nan
        )


        balanced_accuracy = (

            balanced_accuracy_score(
                y_true,
                y_pred
            )

            if (
                len(
                    np.unique(
                        y_true
                    )
                )
                == 2
            )

            else np.nan
        )


        f1 = (

            f1_score(
                y_true,
                y_pred,
                zero_division=0
            )

            if actual_dropouts > 0

            else np.nan
        )


        roc_auc = (

            roc_auc_score(
                y_true,
                y_prob
            )

            if (
                len(
                    np.unique(
                        y_true
                    )
                )
                == 2
            )

            else np.nan
        )


        pr_auc = (

            average_precision_score(
                y_true,
                y_prob
            )

            if actual_dropouts > 0

            else np.nan
        )


        mean_predicted_risk = (
            y_prob.mean()
        )


        calibration_gap = (
            mean_predicted_risk
            -
            dropout_prevalence
        )


        results.append({

            "group":
                group_value,

            "n":
                n,

            "actual_dropouts":
                actual_dropouts,

            "actual_graduates":
                actual_graduates,

            "dropout_prevalence":
                dropout_prevalence,

            "alert_rate":
                alert_rate,

            "precision":
                precision,

            "recall":
                recall,

            "false_negative_rate":
                false_negative_rate,

            "specificity":
                specificity,

            "false_positive_rate":
                false_positive_rate,

            "accuracy":
                accuracy,

            "balanced_accuracy":
                balanced_accuracy,

            "f1":
                f1,

            "roc_auc":
                roc_auc,

            "pr_auc":
                pr_auc,

            "mean_predicted_risk":
                mean_predicted_risk,

            "calibration_gap":
                calibration_gap,

            "true_positives":
                tp,

            "false_negatives":
                fn,

            "true_negatives":
                tn,

            "false_positives":
                fp
        })


    result = pd.DataFrame(
        results
    )


    result.insert(
        0,
        "audit_dimension",
        group_column
    )


    return result
```

### Mark Groups with Sufficient Support

A subgroup with only a handful of dropouts or graduates can produce unstable recall or false-positive rates.

Rather than pretending every number is equally reliable, we will identify groups with reasonable support.

```python
# ============================================================
# MARK GROUPS WITH SUFFICIENT SUPPORT
# ============================================================

MIN_GROUP_SIZE = 30

MIN_POSITIVE_CASES = 10

MIN_NEGATIVE_CASES = 10


for dimension in fairness_tables:

    table = fairness_tables[
        dimension
    ].copy()


    table[
        "sufficient_support"
    ] = (

        (
            table["n"]
            >= MIN_GROUP_SIZE
        )

        &

        (
            table[
                "actual_dropouts"
            ]
            >= MIN_POSITIVE_CASES
        )

        &

        (
            table[
                "actual_graduates"
            ]
            >= MIN_NEGATIVE_CASES
        )
    )


    fairness_tables[
        dimension
    ] = table
```

### Calculate Disparity Gaps

The most useful comparisons here are:

* Recall gap → equal-opportunity diagnostic
* False-negative-rate gap → missed-support disparity
* False-positive-rate gap → unnecessary-alert disparity
* Precision gap
* Alert-rate gap
* Calibration spread

```python
# ============================================================
# CALCULATE SUBGROUP DISPARITY GAPS
# ============================================================

gap_results = []


for dimension, table in fairness_tables.items():

    eligible = table[
        table[
            "sufficient_support"
        ]
    ].copy()


    if len(eligible) < 2:
        continue


    recall_gap = (
        eligible["recall"].max()
        -
        eligible["recall"].min()
    )


    fnr_gap = (
        eligible[
            "false_negative_rate"
        ].max()
        -
        eligible[
            "false_negative_rate"
        ].min()
    )


    fpr_gap = (
        eligible[
            "false_positive_rate"
        ].max()
        -
        eligible[
            "false_positive_rate"
        ].min()
    )


    precision_gap = (
        eligible["precision"].max()
        -
        eligible["precision"].min()
    )


    alert_rate_gap = (
        eligible["alert_rate"].max()
        -
        eligible["alert_rate"].min()
    )


    calibration_gap_range = (
        eligible["calibration_gap"].max()
        -
        eligible["calibration_gap"].min()
    )


    gap_results.append({

        "audit_dimension":
            dimension,

        "eligible_groups":
            len(eligible),

        "recall_gap":
            recall_gap,

        "false_negative_rate_gap":
            fnr_gap,

        "false_positive_rate_gap":
            fpr_gap,

        "precision_gap":
            precision_gap,

        "alert_rate_gap":
            alert_rate_gap,

        "calibration_gap_range":
            calibration_gap_range
    })


fairness_gap_summary = (
    pd.DataFrame(
        gap_results
    )
    .sort_values(
        "recall_gap",
        ascending=False
    )
    .reset_index(drop=True)
)


fairness_gap_summary.round(4)
```

Subgroup Disparity Analysis. After applying minimum subgroup-support criteria, the largest differences in dropout recall were observed for financial-risk status (20.1 percentage points), debtor status (16.2 points), age (15.1 points), and scholarship status (11.5 points). Smaller recall disparities were observed for displacement status (4.9 points) and gender (3.6 points).

These differences should not automatically be interpreted as discriminatory model behavior. Several audited dimensions, particularly financial-risk and debtor status, were associated with substantial differences in underlying dropout prevalence. Correspondingly, large alert-rate differences often reflected genuine variation in observed outcome rates.

Nevertheless, recall disparities are operationally important because they measure the likelihood that an actual dropout will be detected by the early-warning system. Age warrants particular attention because all five age groups had adequate sample support and the dimension exhibited substantial disparities in recall, precision, and calibration. Scholarship status also showed a meaningful recall gap.

Financial-risk variables produced the largest recall differences overall, indicating that the model identifies students with visible financial difficulties considerably more reliably than students whose eventual dropout is not accompanied by those signals. This suggests that the model may perform best for risk pathways represented in institutional administrative data and less effectively for latent or unmeasured causes of withdrawal.

The audit therefore supports continued subgroup monitoring, periodic recalibration, and use of the model as a decision-support tool rather than an exclusive gatekeeper for student assistance.

### Identify potentially vulnerable subgroups

Identify the lowest-recall and highest-FPR groups. This produces a very useful diagnostic summary.

```python
# ============================================================
# IDENTIFY POTENTIALLY VULNERABLE SUBGROUPS
# ============================================================

vulnerability_rows = []


for dimension, table in fairness_tables.items():

    eligible = table[
        table[
            "sufficient_support"
        ]
    ].copy()


    if len(eligible) == 0:
        continue


    lowest_recall_row = (
        eligible
        .sort_values(
            "recall",
            ascending=True
        )
        .iloc[0]
    )


    highest_fpr_row = (
        eligible
        .sort_values(
            "false_positive_rate",
            ascending=False
        )
        .iloc[0]
    )


    vulnerability_rows.append({

        "audit_dimension":
            dimension,

        "lowest_recall_group":
            lowest_recall_row[
                "group"
            ],

        "lowest_recall":
            lowest_recall_row[
                "recall"
            ],

        "lowest_recall_n":
            lowest_recall_row[
                "n"
            ],

        "highest_fpr_group":
            highest_fpr_row[
                "group"
            ],

        "highest_fpr":
            highest_fpr_row[
                "false_positive_rate"
            ],

        "highest_fpr_n":
            highest_fpr_row[
                "n"
            ]
    })


fairness_vulnerability_summary = (
    pd.DataFrame(
        vulnerability_rows
    )
)


fairness_vulnerability_summary.round(4)
```

Accordingly, future monitoring should separately track false-negative and false-positive disparities rather than relying only on aggregate accuracy.

Tuition status requires additional caution because only one subgroup met the predefined class-support criteria. The available metrics remain informative for operational monitoring, but there is insufficient supported evidence for a stable two-group fairness comparison.

### Calibration Diagnostic

This checks whether average predicted risk roughly corresponds to actual dropout prevalence within each group.

A positive calibration gap means the model's average predicted risk is higher than the observed dropout rate; a negative gap means it is lower.

```python
# ============================================================
# SUBGROUP CALIBRATION SUMMARY
# ============================================================

calibration_rows = []


for dimension, table in fairness_tables.items():

    for _, row in table.iterrows():

        calibration_rows.append({

            "audit_dimension":
                dimension,

            "group":
                row["group"],

            "n":
                row["n"],

            "observed_dropout_rate":
                row[
                    "dropout_prevalence"
                ],

            "mean_predicted_risk":
                row[
                    "mean_predicted_risk"
                ],

            "calibration_gap":
                row[
                    "calibration_gap"
                ],

            "sufficient_support":
                row[
                    "sufficient_support"
                ]
        })


fairness_calibration = (
    pd.DataFrame(
        calibration_rows
    )
)


display(
    fairness_calibration[
        fairness_calibration[
            "sufficient_support"
        ]
    ]
    .sort_values(
        "calibration_gap",
        key=lambda series:
            series.abs(),
        ascending=False
    )
    .round(4)
)
```

Subgroup Calibration Audit. Average predicted dropout probabilities were generally well aligned with observed subgroup dropout prevalence. Among adequately supported groups, calibration gaps remained within approximately ±6 percentage points. The largest overestimation occurred for Displaced code 1 (+5.7 percentage points), followed by students aged 21–24 (+5.5 points) and Scholarship code 1 (+5.1 points).

Most other groups exhibited smaller discrepancies, and slight underprediction was observed only for the 25–29 and 30–39 age groups. Overall, the calibration results do not indicate severe subgroup-level probability distortion.

Importantly, reasonable calibration did not necessarily imply equal classification performance. Some subgroups with modest calibration gaps still exhibited substantial recall disparities at the frozen 0.48 threshold. This demonstrates that probability calibration and threshold-based fairness measure different aspects of model behavior and should be monitored separately.

#### Record Human Override

```python
# ============================================================
# RECORD HUMAN OVERRIDE
# ============================================================

def record_override(
    override_log,
    case_id,
    prediction_timestamp,
    model_version,
    decision_threshold,
    predicted_dropout_probability,
    model_alert,
    human_final_decision,
    override_reason_category,
    override_reason_notes,
    reviewer_role,
    follow_up_action,
    review_timestamp
):

    if override_reason_category not in override_reason_categories:

        raise ValueError(
            "override_reason_category must be one of the "
            "approved override reason categories."
        )


    if (
        model_alert == 1
        and human_final_decision == 0
    ):

        override_type = (
            "Model alert not actioned"
        )


    elif (
        model_alert == 0
        and human_final_decision == 1
    ):

        override_type = (
            "Human-initiated intervention"
        )


    else:

        override_type = (
            "No override"
        )


    new_record = pd.DataFrame(
        [
            {
                "case_id":
                    case_id,

                "prediction_timestamp":
                    prediction_timestamp,

                "model_version":
                    model_version,

                "decision_threshold":
                    decision_threshold,

                "predicted_dropout_probability":
                    predicted_dropout_probability,

                "model_alert":
                    model_alert,

                "human_final_decision":
                    human_final_decision,

                "override_type":
                    override_type,

                "override_reason_category":
                    override_reason_category,

                "override_reason_notes":
                    override_reason_notes,

                "reviewer_role":
                    reviewer_role,

                "follow_up_action":
                    follow_up_action,

                "review_timestamp":
                    review_timestamp
            }
        ]
    )


    return pd.concat(
        [
            override_log,
            new_record
        ],
        ignore_index=True
    )
```

```python
# ============================================================
# FORMAL FAIRNESS METRICS
# ============================================================

formal_fairness_rows = []


for audit_dimension, table in fairness_tables.items():

    eligible = (
        table[
            table["sufficient_support"] == True
        ]
        .copy()
    )


    if len(eligible) < 2:

        formal_fairness_rows.append(
            {
                "audit_dimension":
                    audit_dimension,

                "eligible_groups":
                    len(eligible),

                "demographic_parity_difference":
                    np.nan,

                "disparate_impact_ratio":
                    np.nan,

                "equal_opportunity_difference":
                    np.nan,

                "equalised_odds_difference":
                    np.nan,

                "tpr_gap":
                    np.nan,

                "fpr_gap":
                    np.nan,

                "assessment_status":
                    "Insufficient comparable support"
            }
        )

        continue


    alert_rates = (
        eligible["alert_rate"]
        .astype(float)
    )


    tpr_values = (
        eligible["recall"]
        .astype(float)
    )


    fpr_values = (
        eligible["false_positive_rate"]
        .astype(float)
    )


    demographic_parity_difference = (
        alert_rates.max()
        -
        alert_rates.min()
    )


    if alert_rates.max() > 0:

        disparate_impact_ratio = (
            alert_rates.min()
            /
            alert_rates.max()
        )

    else:

        disparate_impact_ratio = np.nan


    tpr_gap = (
        tpr_values.max()
        -
        tpr_values.min()
    )


    fpr_gap = (
        fpr_values.max()
        -
        fpr_values.min()
    )


    equal_opportunity_difference = (
        tpr_gap
    )


    equalised_odds_difference = max(
        tpr_gap,
        fpr_gap
    )


    formal_fairness_rows.append(
        {
            "audit_dimension":
                audit_dimension,

            "eligible_groups":
                len(eligible),

            "demographic_parity_difference":
                demographic_parity_difference,

            "disparate_impact_ratio":
                disparate_impact_ratio,

            "equal_opportunity_difference":
                equal_opportunity_difference,

            "equalised_odds_difference":
                equalised_odds_difference,

            "tpr_gap":
                tpr_gap,

            "fpr_gap":
                fpr_gap,

            "assessment_status":
                "Comparable"
        }
    )


formal_fairness_metrics = (
    pd.DataFrame(
        formal_fairness_rows
    )
    .sort_values(
        "equalised_odds_difference",
        ascending=False,
        na_position="last"
    )
    .reset_index(drop=True)
)


display(
    formal_fairness_metrics.round(4)
)
```

Formal Fairness Metrics. Formal subgroup fairness measures were evaluated using only groups satisfying the predefined minimum class-support criteria. Demographic parity was assessed through differences in model alert rates, while disparate impact was summarized as the ratio of the lowest to highest supported subgroup alert rate. Equal opportunity was evaluated using differences in true-positive rate (dropout recall), and equalised odds considered the larger disparity across true-positive and false-positive rates.

The largest equal-opportunity and equalised-odds differences were observed for financial-risk status (20.1 percentage points), debtor status (16.2 points), age (15.1 points), and scholarship status (11.5 points). Displacement showed a smaller difference of 4.9 points. Gender exhibited relatively similar dropout recall across groups, but its equalised-odds difference increased to 6.0 points because the false-positive-rate disparity exceeded the recall disparity.

Demographic-parity differences were substantially larger because alert rates also reflected major differences in underlying dropout prevalence across the audited populations. Accordingly, demographic parity and disparate-impact ratios were interpreted as descriptive screening measures rather than standalone evidence of discriminatory model behavior. For this early-warning application, equal opportunity and equalised odds provide more direct evidence regarding differences in model error behavior.

Tuition-status groups could not support a formal comparative fairness assessment because only one subgroup met the predefined positive- and negative-class support requirements.

```python
# ============================================================
# PREPARE RAW DATA FOR LIME
# ============================================================

import numpy as np
import pandas as pd

from scipy import sparse


final_pipeline = (
    xgb_search.best_estimator_
)


if hasattr(
    final_pipeline,
    "feature_names_in_"
):

    expected_input_columns = list(
        final_pipeline.feature_names_in_
    )

else:

    expected_input_columns = list(
        final_pipeline.steps[0][1].feature_names_in_
    )


# Use the validated training dataset from the PDP analysis
X_train_lime_raw = (
    X_pdp[
        expected_input_columns
    ]
    .copy()
)


# Find the untouched holdout DataFrame
holdout_candidates = []


for name, obj in list(
    globals().items()
):

    if not isinstance(
        obj,
        pd.DataFrame
    ):
        continue


    if len(obj) != len(y_test):
        continue


    if not set(
        expected_input_columns
    ).issubset(
        obj.columns
    ):
        continue


    index_match = False

    if hasattr(
        y_test,
        "index"
    ):

        index_match = (
            obj.index.equals(
                y_test.index
            )
        )


    holdout_candidates.append(
        {
            "name":
                name,

            "data":
                obj,

            "index_match":
                index_match,

            "column_count":
                len(obj.columns)
        }
    )


if len(
    holdout_candidates
) == 0:

    raise ValueError(
        "No holdout DataFrame matching the fitted model "
        "input schema was found."
    )


holdout_candidates = sorted(
    holdout_candidates,
    key=lambda item: (
        not item["index_match"],
        item["column_count"],
        item["name"]
    )
)


holdout_source_name = (
    holdout_candidates[0][
        "name"
    ]
)


holdout_source = (
    holdout_candidates[0][
        "data"
    ]
)


X_test_lime_raw = (
    holdout_source[
        expected_input_columns
    ]
    .copy()
)


print(
    "Training shape:",
    X_train_lime_raw.shape
)

print(
    "Holdout source selected:",
    holdout_source_name
)

print(
    "Holdout shape:",
    X_test_lime_raw.shape
)

print(
    "Expected raw features:",
    len(
        expected_input_columns
    )
)
```

#### LIME Local Explanation

```python
# ============================================================
# LIME LOCAL EXPLANATION
# ============================================================

from lime.lime_tabular import LimeTabularExplainer


final_probabilities = (
    final_pipeline
    .predict_proba(
        X_test_lime_raw
    )[:, 1]
)


final_predictions = (
    final_probabilities
    >=
    final_threshold
).astype(int)


y_test_array = np.asarray(
    y_test
)


true_positive_positions = np.where(
    (
        y_test_array == 1
    )
    &
    (
        final_predictions == 1
    )
)[0]


tp_probabilities = (
    final_probabilities[
        true_positive_positions
    ]
)


median_order = np.argsort(
    tp_probabilities
)


representative_position = int(
    true_positive_positions[
        median_order[
            len(
                median_order
            )
            //
            2
        ]
    ]
)


lime_explainer = LimeTabularExplainer(
    training_data=X_train_lime,
    feature_names=lime_feature_names,
    class_names=[
        "Graduate",
        "Dropout"
    ],
    mode="classification",
    discretize_continuous=True,
    random_state=42
)


dropout_class_index = list(
    final_estimator.classes_
).index(
    1
)


lime_explanation = (
    lime_explainer
    .explain_instance(
        X_test_lime[
            representative_position
        ],
        final_estimator.predict_proba,
        labels=[
            dropout_class_index
        ],
        num_features=12
    )
)


lime_result = pd.DataFrame(
    lime_explanation.as_list(
        label=dropout_class_index
    ),
    columns=[
        "feature_condition",
        "lime_weight"
    ]
)


lime_result[
    "absolute_weight"
] = (
    lime_result[
        "lime_weight"
    ]
    .abs()
)


lime_result[
    "direction"
] = np.where(
    lime_result[
        "lime_weight"
    ]
    > 0,
    "Increases predicted dropout risk",
    "Decreases predicted dropout risk"
)


print(
    "Representative holdout position:",
    representative_position
)

print(
    "Actual class:",
    y_test_array[
        representative_position
    ]
)

print(
    "Predicted dropout probability:",
    round(
        final_probabilities[
            representative_position
        ],
        4
    )
)

print(
    "Frozen decision threshold:",
    final_threshold
)


display(
    lime_result[
        [
            "feature_condition",
            "lime_weight",
            "absolute_weight",
            "direction"
        ]
    ]
    .round(4)
)
```

# Bias & Fairness Analysis

## 1. Purpose and Scope

A bias and fairness assessment was conducted for the final XGBoost student-dropout prediction model to determine whether strong aggregate predictive performance was accompanied by materially different behavior across student subgroups. The analysis also examined model explainability, potential data leakage, overfitting, subgroup calibration, false-negative risk, and operational safeguards.

The final model uses a frozen 0.48 decision threshold and predicts dropout risk at the end of Semester 1. All final fairness analyses were performed on the 726-record untouched holdout set. The model itself and its operating threshold were not modified based on the fairness results, preserving the independence of the final evaluation.

The final XGBoost model achieved an accuracy of 0.8939, balanced accuracy of 0.8902, precision of 0.8581, recall of 0.8732, F1-score of 0.8656, ROC-AUC of 0.9540, and PR-AUC of 0.9492. Despite this strong overall performance, the model produced 36 false negatives and 41 false positives, demonstrating that aggregate performance alone is insufficient for evaluating responsible use.

## 2. Explainability Analysis

Multiple complementary explainability techniques were used to understand both global model behavior and individual predictions: SHAP, LIME, Partial Dependence Plots (PDP), and Individual Conditional Expectation (ICE).

### 2.1 Global SHAP Analysis

SHAP analysis was validated through probability reconstruction, with a maximum reconstruction error of approximately 3.58 × 10⁻⁷, confirming that the calculated SHAP values accurately reconstructed the model's predicted probabilities.

The strongest global predictor by a substantial margin was Semester 1 completion rate. Other influential variables included tuition-payment status, financial-risk count, Semester 1 evaluation success rate, number of curricular units enrolled, scholarship status, Semester 1 grade, age at enrollment, course, and unemployment rate.

This feature ranking indicates that the model relies heavily on academic engagement and observable financial or administrative signals. While these variables are relevant to dropout prediction, their influence creates an important governance consideration: students whose difficulties are not captured through these institutional indicators may be less visible to the model.

Local SHAP analysis reinforced this point. In representative correctly identified dropout cases, financial and academic signals often increased predicted dropout risk. Analysis of false-negative cases demonstrated that combinations of apparently protective characteristics could outweigh other risk factors, causing some actual dropouts to remain below the 0.48 alert threshold.

### 2.2 Partial Dependence

The partial dependence analysis for sem1_completion_rate showed a strong nonlinear inverse association with predicted dropout risk. Predicted risk remained high at low completion rates and declined markedly as completion increased, with especially pronounced reductions through the middle-to-high completion range.

This finding is consistent with the global SHAP results and confirms that Semester 1 completion rate is a dominant feature in the fitted XGBoost model. Because XGBoost is tree based, the step-like pattern observed in the PDP is expected.

The relationship should not be interpreted causally. The analysis demonstrates how the model responds to completion rate while averaging over the remaining observed features; it does not establish that changing completion rate alone would cause the corresponding change in actual dropout probability.

### 2.3 Individual Conditional Expectation

ICE analysis showed that the general inverse relationship between Semester 1 completion and dropout risk also appears at the individual level, but the magnitude and timing of the effect vary considerably across students.

For many students, predicted risk decreased sharply when completion entered approximately the 60%–85% range. For other students, risk remained relatively high even at high completion rates, indicating that financial, demographic, course-related, or other academic variables continued to influence the prediction.

The ICE results therefore demonstrate that Semester 1 completion is highly influential but does not operate independently of the broader student profile.

### 2.4 LIME Analysis

LIME was used as an additional local explanation technique. A representative true-positive holdout case had an actual dropout outcome and a predicted dropout probability of 0.9708, substantially above the frozen 0.48 threshold.

LIME identified a low transformed Semester 1 completion value as a contributor toward increased predicted dropout risk, supporting the direction observed in SHAP, PDP, and ICE. Several one-hot encoded categorical indicators also contributed to the local explanation.

Because LIME operated in the 228-feature transformed model space, some conditions refer to the presence or absence of encoded categories rather than directly interpretable raw variables. These coefficients describe the local surrogate model and should not be interpreted as direct percentage-point changes in dropout probability. For substantive interpretation, SHAP and raw-feature analyses were therefore treated as the primary explanation methods.

## 3. Subgroup Fairness Audit

The final model was audited across the following primary dimensions: gender, age, displacement status, scholarship status, debtor status, tuition-payment status, and financial-risk count.

For each eligible subgroup, the analysis evaluated sample size, dropout prevalence, alert rate, precision, recall, false-negative rate, false-positive rate, specificity, balanced accuracy, mean predicted risk, and calibration gap.

Minimum class-support rules were applied before making formal subgroup comparisons. Groups without adequate positive and negative outcome support were retained for descriptive monitoring but excluded from formal disparity conclusions. This was particularly relevant for some tuition-status and financial-risk categories.

The subgroup results showed that predictive performance was not uniform across the audited populations. The largest recall disparities occurred for financial-risk status, debtor status, age, and scholarship status. Smaller recall differences were observed for displacement and gender.

## 4. Formal Fairness Metrics

Four formal fairness concepts were evaluated.

Demographic parity difference measures the range in model alert rates across supported groups.
Disparate impact ratio was calculated as the lowest supported subgroup alert rate divided by the highest supported subgroup alert rate.
Equal opportunity difference measures the range in true-positive rate, equivalent here to dropout recall.
Equalised odds difference captures the larger of the subgroup true-positive-rate and false-positive-rate differences.

Tuition-status groups did not provide sufficient comparable support for a formal two-group fairness assessment, so no formal parity or equalised-odds conclusion was drawn for that dimension.

## 5. Interpretation of Demographic Parity and Disparate Impact

The largest demographic-parity differences were substantially greater than the corresponding equal-opportunity differences. This is important because the audited subgroups also exhibited large differences in observed dropout prevalence.

For example, students with visible financial-risk or debtor indicators had materially higher observed dropout rates than comparison groups. A model that successfully ranks dropout risk will therefore naturally produce different alert rates when underlying outcome prevalence differs.

For this reason, demographic parity and the reported disparate-impact ratios were treated as screening measures rather than standalone evidence of unfairness. The disparate-impact calculation used in this analysis is a lowest-to-highest subgroup alert-rate ratio and should not be interpreted as a legal determination or universal pass/fail test.

For this early-warning use case, equal opportunity and equalised odds are more operationally informative because they directly evaluate whether the model makes different types of errors among students with known outcomes.

## 6. Subgroups Requiring Enhanced Monitoring

Fairness-monitoring priorities were ranked primarily according to dropout-recall disparity because a false negative represents a student who actually drops out but is not flagged for proactive support.

The monitoring priority order was financial-risk group, debtor status, age, scholarship status, displacement status, and gender.

Financial-risk status showed the largest recall gap at 20.11 percentage points. Debtor status followed at 16.22 points, age at 15.12 points, and scholarship status at 11.45 points. Age also warrants particular attention because it exhibited the largest precision disparity and the largest range in subgroup calibration.

These rankings are intended to prioritize monitoring effort; they are not overall fairness scores and do not establish discriminatory treatment.

## 7. Calibration

Subgroup calibration was evaluated by comparing observed dropout prevalence with mean predicted dropout risk.

Calibration was generally reasonably close for many groups, but some groups showed larger deviations. The most notable supported calibration gaps were observed for displaced students, younger age categories, and scholarship recipients.

Age also showed the widest calibration-gap range across supported groups. This reinforces the need to monitor not only classification errors but also whether predicted probabilities retain similar meaning across populations.

Calibration findings should be re-evaluated on future cohorts because calibration can shift even when ranking metrics such as ROC-AUC remain stable.

## 8. Model Limitations

The bias and fairness analysis identified several limitations that must accompany interpretation of the final model.

The holdout dropout prevalence was approximately 39.1%, representing moderate rather than extreme class imbalance. Consequently, model quality was evaluated using PR-AUC, recall, F1, F2, balanced accuracy, ROC-AUC, precision, specificity, and confusion-matrix measures rather than accuracy alone.

Temporal leakage is an important risk because Semester 1 completion and academic-performance variables are among the model's strongest predictors. The operational prediction horizon is therefore explicitly fixed at End of Semester 1. The model should not be applied earlier using variables that would not yet genuinely be available.

Train-holdout leakage was separately checked, with zero index overlap confirmed between the 2,904 training records and 726 untouched holdout records.

XGBoost's nonlinear flexibility creates potential overfitting risk. However, cross-validation, out-of-fold analysis, and final holdout evaluation indicated reasonable generalization. Strong holdout ROC-AUC and PR-AUC results further support this conclusion, although they do not eliminate the need for monitoring on future cohorts.

The final model still produced 36 false negatives. A non-alert must therefore never be interpreted as confirmation that a student has no risk. Alternative pathways such as faculty referral, advisor identification, student self-referral, and other support mechanisms must remain available.

Subgroup sample-size constraints also limit some fairness conclusions. Small positive or negative class counts can make subgroup metrics unstable, so formal conclusions were restricted to groups satisfying predefined support requirements.

The model also depends heavily on observable institutional academic and financial signals. Students experiencing difficulties outside those recorded systems may be less likely to receive model-generated alerts.

Finally, SHAP, LIME, PDP, and ICE explain predictive behavior rather than causal relationships. Feature contributions must not be interpreted as proof that manipulating a variable will cause the corresponding change in student outcomes.

## 9. Bias Mitigation and Responsible Use

The fairness findings do not justify modifying the final model or threshold using the final holdout set. Doing so would convert the holdout into an additional development dataset and weaken the integrity of the evaluation.

Instead, the current model remains frozen, while mitigation is divided into operational safeguards and future model-development experiments.

Operational controls include human review of alerts, documented human overrides, alternative referral pathways for non-alerted students, subgroup fairness monitoring, program-level performance monitoring, probability-calibration monitoring, privacy and access controls, and audit logging with model-version tracking.

The model is explicitly prohibited from independently determining withdrawal, dismissal, admission, scholarship denial, disciplinary action, or denial of institutional services. Its approved role is to identify potential opportunities for additional support.

Future mitigation experiments should be conducted using new development and validation data. These include feature-ablation analysis, financial-feature sensitivity testing, false-negative feature discovery, probability recalibration, threshold sensitivity analysis, comparison with alternative models, and program-specific diagnostics. Reweighting, threshold modification, or other fairness interventions should be evaluated only in a future controlled development cycle rather than retrofitted to the current holdout results.

## 10. Responsible AI Assessment

The final model demonstrates strong predictive discrimination, balanced classification performance, and technically validated explainability. At the same time, the fairness audit identifies meaningful differences in error distribution across several student populations, particularly financial-risk, debtor, age, and scholarship groups.

The results therefore do not support a simple conclusion that the model is universally “fair” or “unfair.” Instead, they show that model performance is strong overall but heterogeneous across subgroups, requiring contextual interpretation and continuing oversight.

The final deployment position is:

**CONDITIONAL GO — SUPPORTIVE DECISION SUPPORT ONLY**

The XGBoost model and 0.48 operating threshold remain frozen. The model may support prioritization of academic advising, financial-support outreach, counseling, and other beneficial interventions, but it must remain subject to human judgment and must not independently trigger adverse consequences.

The governing principle is:

**A high-risk prediction may create an additional opportunity for support; a low-risk prediction must never remove a student's opportunity to receive support.**

Future mitigation, recalibration, threshold adjustment, or model redevelopment must use new development and validation data, while the current untouched holdout remains preserved as the final evaluation benchmark.
